# 02 Data cleaning
In ***01_explore.ipynb*** I looked at the raw data and decided what needs to change. In this notebook I put these decisions into code. The result is ***one clean table***, saved in ***data/processed/***.
The raw file in ***data/raw/ffa/*** stays unchanged.

```python
# load the paths
from pathlib import Path  # to work with file paths
import pandas as pd

DATA_FILE = Path('../../data/raw/ffa/ffa_tickets_total_arthouse_2001_2025.csv')  # raw data (input)
OUT_FILE = Path('../../data/processed/q3_ffa_tickets.csv')                     # clean data (output)

DATA_FILE.exists()  # if True = the file was found
```

```python
# loading the data
dataset = pd.read_csv(DATA_FILE)
```

```python
# removing columns that are not needed for the analysis
dataset = dataset.drop(columns=['tickets_arthouse_million', 'arthouse_definition', 'source_id', 'source'])

dataset.shape  # expected: 6 - 4 = 2 columns
```

```python
# check for missing values
dataset.isna().sum()  # counts missing values per column, 0 = nothing missing
```

```python
# save the clean table
dataset.to_csv(OUT_FILE, index=False)
```

```python
# reload and check the saved file
check = pd.read_csv(OUT_FILE)
check
```

### Summary
**Result:** `data/processed/q3_ffa_tickets.csv` with 2 columns and 25 rows (2001–2025).

**What changed:** only four columns were removed, because they are not necessary for our project and analysis.

| Removed column | Why |
|---|---|
| `tickets_arthouse_million` | We only need the total number of tickets |
| `arthouse_definition` | Only describes the arthouse numbers, which we removed |
| `source_id` | Source is documented in the README |
| `source` | Source is documented in the README |

**Not changed on purpose:** `year` and `tickets_total_million`. These are the only columns we need, and they have no missing values.
